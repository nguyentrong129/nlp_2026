# LAB 09 — Tokenizer Alignment
Quan sát word-level BIO labels được căn chỉnh với subword tokens. Strategy: nhãn cho subword đầu tiên, phần còn lại IGNORE (-100).

```python
# !pip -q install transformers pandas
from transformers import AutoTokenizer
import pandas as pd
tok=AutoTokenizer.from_pretrained("bert-base-cased")
words=["Barack","Obama","visited","New","York","."]
labels=["B-PER","I-PER","O","B-LOC","I-LOC","O"]
enc=tok(words,is_split_into_words=True)
rows=[]; prev=None
for sub,wid in zip(tok.convert_ids_to_tokens(enc["input_ids"]),enc.word_ids()):
    lab="IGNORE" if wid is None or wid==prev else labels[wid]
    rows.append({"subword":sub,"word_id":wid,"word":words[wid] if wid is not None else None,"BIO":lab})
    prev=wid
display(pd.DataFrame(rows))
```

Thử thay bằng một từ dài/ít phổ biến để quan sát subword splitting. Tự giải thích vì sao các subword tiếp theo bị gán IGNORE.
